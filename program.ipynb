```python
import cv2
from ultralytics import YOLO


```

```python
model = YOLO("yolov8n.pt")
face_model = YOLO("yolov8m-face.pt")
```

```python
photo = cv2.imread('demo.jpg')

```

```python
result= model(photo)
face_result = face_model(photo)



```

```python
person = result[0].plot()
face = face_result[0].plot(img=person)
```

```python
cv2.imshow("my window" , face)
cv2.moveWindow("my window" , 1800 , 200)
```

```python
cap = cv2.VideoCapture(0)


while cv2.waitKey(1) != ord("x"):
    _, frame = cap.read()

   
    result = model(frame, verbose=False)
    face_result = face_model(frame, verbose=False)

    person = result[0].plot()
    face = face_result[0].plot(img=person)
    
    cv2.imshow("my window", face)

    cv2.moveWindow("my window", 1800, 200)

cv2.waitKey(5000)
cap.release()
cv2.destroyAllWindows()
```
